```python

from zipfile import ZipFile
from pathlib import Path

extracted_files = []

with ZipFile(ZIP_PATH, "r") as archive:

    for member in csv_files:

        # Preserve the original internal folder structure
        relative_path = Path(member)

        destination = EXTRACTED_DIR / relative_path

        # Prevent unsafe archive paths
        destination = destination.resolve()
        destination.relative_to(EXTRACTED_DIR.resolve())

        destination.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        # Extract without changing file contents
        destination.write_bytes(archive.read(member))

        extracted_files.append(destination)

print("DATASET EXTRACTION")
print("=" * 55)

print("Files extracted:", len(extracted_files))
print("Extraction directory:", EXTRACTED_DIR)

print("\nEXTRACTED FILES:")

for file in extracted_files:
    print(
        f"{file.name} | "
        f"{file.stat().st_size / 1024:.2f} KB"
    )
```

```python

import pandas as pd

extraction_results = []

for file in extracted_files:

    extraction_results.append({
        "file_name": file.name,
        "exists": file.exists(),
        "size_bytes": file.stat().st_size,
        "is_empty": file.stat().st_size == 0
    })

extraction_df = pd.DataFrame(extraction_results)

print("EXTRACTION VERIFICATION")
print("=" * 55)

print("Expected files:", len(csv_files))
print("Extracted files:", len(extracted_files))
print("Files existing:", extraction_df["exists"].sum())
print("Empty files:", extraction_df["is_empty"].sum())

print(
    "\nAll files verified:",
    len(extracted_files) == len(csv_files)
    and extraction_df["exists"].all()
    and not extraction_df["is_empty"].any()
)

display(extraction_df)
```
